# Word Chain Game.

```python
import random
import time

def play_word_chain():
    print("Welcome to the Word Chain Game!")
    print("Rules:")
    print("1. The AI starts with a random word.")
    print("2. You must enter a word that starts with the LAST letter of AI's word.")
    print("3. No repeating words! Type 'quit' to give up.")
    print("4. Each correct answer = +1 point.\n")

    # Difficulty level
    level = input("Choose difficulty (easy / hard): ").strip().lower()
    if level == "easy":
        words = ["apple", "egg", "rose", "ear", "rat", "eye", "orange", "tree", "ice", "ant"]
    else:
        words = ["elephant", "tiger", "eagle", "tree", "goat", "orange", "ant", "ear", "rain", "night",
                 "yak", "kangaroo", "octopus", "snake", "dog", "grape", "eggplant", "lemon"]

    # AI starts the game
    ai_word = random.choice(words)
    used_words = [ai_word]
    score = 0

    print(f"\nAI starts with: {ai_word}")

    # Game loop
    while True:
        last_letter = ai_word[-1]
        user_word = input(f"Your turn! Enter a word starting with '{last_letter}': ").lower().strip()

        # Player quits
        if user_word == "quit":
            print("\nYou gave up! AI wins!")
            break

        # Check if the word was already used
        if user_word in used_words:
            print("That word was already used! AI wins!")
            break

        # Check if the word starts with the correct letter
        if user_word[0] != last_letter:
            print(f"Wrong letter! Your word must start with '{last_letter}'. AI wins!")
            break

        # Check if the word exists in our list (optional check)
        if user_word not in words:
            print("Unknown word! Try a valid one next time.")
            break

        # If valid word → continue
        used_words.append(user_word)
        score += 1
        print(f"Good job! Current score: {score}")

        # AI's turn — find a word starting with the last letter of user's word
        next_letter = user_word[-1]
        possible_words = [w for w in words if w.startswith(next_letter) and w not in used_words]

        time.sleep(1)

        if possible_words:
            ai_word = random.choice(possible_words)
            used_words.append(ai_word)
            ai_comments = ["Nice move!", "I'm still in!", "You’re good!", "Haha, I got one!"]
            print(f"AI: {ai_word} — {random.choice(ai_comments)}")
        else:
            print(f"\nAI can't think of a word starting with '{next_letter}' — YOU WIN!")
            print(f"Final Score: {score}")
            break

    print("\nGame Over! Thanks for playing Word Chain\n")

# Main loop to replay the game
if __name__ == "__main__":
    while True:
        play_word_chain()
        again = input("Play again? (y/n): ").strip().lower()
        if again != "y":
            print("\nGame ended. Hope you enjoyed! ")
            break
```

#  Tic Toc Toe Game.

```python
import random

def print_board(board):
    for row in board:
        print(" | ".join(row))
        print("-" * 5)

def check_winner(board, player):
    # Rows, columns, diagonals
    for i in range(3):
        if all(board[i][j] == player for j in range(3)):
            return True
        if all(board[j][i] == player for j in range(3)):
            return True
    if all(board[i][i] == player for i in range(3)):
        return True
    if all(board[i][2-i] == player for i in range(3)):
        return True
    return False

def get_empty_cells(board):
    return [(r, c) for r in range(3) for c in range(3) if board[r][c] == " "]

def make_move(board, move, player):
    board[move[0]][move[1]] = player

def player_move(board):
    while True:
        try:
            move = input("Enter your move (row and column, e.g. 1 2): ")
            r, c = map(int, move.split())
            if board[r-1][c-1] == " ":
                board[r-1][c-1] = "X"
                break
            else:
                print("That cell is already taken!")
        except (ValueError, IndexError):
            print("Invalid input! Please enter row and column between 1-3.")

def find_best_move(board):
    empty = get_empty_cells(board)

    # 1. Try To win
    for move in empty:
        make_move(board, move, "O")
        if check_winner(board, "O"):
            return move
        board[move[0]][move[1]] = " "

    # 2. Block player’s win
    for move in empty:
        make_move(board, move, "X")
        if check_winner(board, "X"):
            board[move[0]][move[1]] = " "
            return move
        board[move[0]][move[1]] = " "

    # 3. Take center if free
    if (1, 1) in empty:
        return (1, 1)

    # 4. Take any corner
    corners = [(0,0), (0,2), (2,0), (2,2)]
    available_corners = [c for c in corners if c in empty]
    if available_corners:
        return random.choice(available_corners)

    # 5. Take any side
    sides = [(0,1), (1,0), (1,2), (2,1)]
    available_sides = [s for s in sides if s in empty]
    if available_sides:
        return random.choice(available_sides)

def computer_move(board):
    move = find_best_move(board)
    if move:
        board[move[0]][move[1]] = "O"
        print(f"Computer chose: {move[0]+1}, {move[1]+1}")

def play_game():
    board = [[" "]*3 for _ in range(3)]
    print("\nWelcome to Tic-Tac-Toe! You are X, computer is O.")
    print_board(board)

    for turn in range(9):
        if turn % 2 == 0:
            player_move(board)
        else:
            computer_move(board)

        print_board(board)

        if check_winner(board, "X"):
            print("You win!")
            return "player"
        elif check_winner(board, "O"):
            print("Computer wins!")
            return "computer"

    print("It's a draw!")
    return "draw"

#Main Loop
while True:
    result = play_game()
    again = input("\nDo you want to play again? (yes/no): ").lower()
    if again not in ["yes", "y"]:
        print("Thanks for playing!")
        break
```
